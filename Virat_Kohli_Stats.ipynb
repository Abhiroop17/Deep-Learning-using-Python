# **Balls Faced**

```python
Balls_faced = df['BF'].sum()
print(f"Balls_faced: {Balls_faced}")
```

# **Average Runs**

```python
average_runs = df['Runs'].mean()
print(f"Average Runs per Innings: {average_runs:.2f}")
```

# **Overall Strike Rate**

```python
overall_strike_rate = (df['Runs'].sum() / df['BF'].sum()) * 100
print(f"Overall Strike Rate: {overall_strike_rate:.2f}")
```

# **Strike Rate Across Innings**

```python
plt.figure(figsize=(10, 6))
plt.bar(df.index, df['SR'], color='red')
plt.xlabel('Match Index')
plt.ylabel('Strike Rate')
plt.title('Strike Rate Across Innings (Scatter Plot)')
plt.show()
```

# **Total Fours**

```python
total_fours = df['4s'].sum()
print(f"Total Fours: {total_fours}")
```

# **Total Sixes**

```python
total_sixes = df['6s'].sum()
print(f"Total Sixes: {total_sixes}")
```

# **Boundary Contribution**

```python
boundary_contribution = ((total_fours * 4 + total_sixes * 6) / total_runs) * 100
print(f"Boundary Contribution to Total Runs: {boundary_contribution:.2f}%")
```

# **Dismissal Counts**

```python
dismissal_counts = df['Dismissal'].value_counts()
print(f"Dismissal Types:\n{dismissal_counts}")
```

# **Distribution of Dismissal Types**

```python
plt.figure(figsize=(10, 6))
dismissal_counts.plot(kind='pie', autopct='%1.1f%%', startangle=140, colors=['lightblue', 'lightgreen', 'orange', 'pink'])
plt.title('Distribution of Dismissal Types')
plt.ylabel('')
plt.show()
```
